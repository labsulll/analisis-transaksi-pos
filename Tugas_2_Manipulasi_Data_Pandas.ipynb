# Task 1 — Indexing & Selection

## 1. Pembersihan Format Harga

Pada bagian ini dilakukan pemeriksaan dan perbaikan tipe data pada kolom harga_satuan. Kolom tersebut masih bertipe object karena terdapat beberapa nilai yang menggunakan koma sebagai pemisah desimal, seperti 7500,00. Koma diubah menjadi titik, kemudian seluruh nilai dikonversi menjadi tipe data float agar dapat digunakan dalam perhitungan numerik.

# Task 2 — Operasi Aritmatika / Feature Engineering

## 1. Membuat Total_Kotor

Pada bagian ini dibuat kolom baru bernama `Total_Kotor` untuk menghitung total harga sebelum diskon. Nilai `Total_Kotor` diperoleh dari perkalian antara `jumlah` barang dengan `harga_satuan`. Hasil perhitungan kemudian disimpan sebagai kolom baru pada DataFrame.

```python
df["Total_Kotor"] = df["jumlah"] * df["harga_satuan"]
```

```python
df[["id_transaksi", "jumlah", "harga_satuan", "Total_Kotor"]].head(10)
```

## 2. Membuat Nominal_Diskon

Pada bagian ini dibuat kolom baru bernama `Nominal_Diskon` untuk menghitung jumlah uang yang mendapatkan potongan harga. Nilai `Nominal_Diskon` dihitung berdasarkan `Total_Kotor` dikalikan dengan persentase diskon pada kolom `diskon_persen`, kemudian dibagi 100.

```python
df["Nominal_Diskon"] = df["Total_Kotor"] * df["diskon_persen"] / 100
```

```python
df[["id_transaksi", "Total_Kotor", "diskon_persen", "Nominal_Diskon"]].head(10)
```

## 3. Perhitungan Omzet Bersih & Pajak

Pada bagian ini dibuat kolom baru bernama `Total_Bersih` untuk menghitung omzet bersih setelah diskon dan ditambahkan pajak PPN sebesar 11%. Perhitungan dilakukan sesuai rumus yang diberikan pada soal, yaitu `(Total_Kotor - Nominal_Diskon) * 1.11`.

```python
df["Total_Bersih"] = (df["Total_Kotor"] - df["Nominal_Diskon"]) * 1.11
```

```python
df[["id_transaksi", "Total_Kotor", "Nominal_Diskon", "Total_Bersih"]].head(10)
```

# Task 3 — Filtering (Penyaringan Data)

## 1. Filter Multi-Kondisi

Pada bagian ini dilakukan penyaringan data berdasarkan beberapa kondisi sekaligus. Data yang ditampilkan harus memiliki status `SELESAI`, berasal dari gerai `G05` atau `G09`, dan memiliki nilai `Total_Bersih` di atas Rp100.000. Setelah proses penyaringan, jumlah baris data yang memenuhi seluruh kriteria akan dihitung.

```python
hasil_filter = df[
    (df["status"] == "SELESAI") &
    (df["kode_gerai"].isin(["G05", "G09"])) &
    (df["Total_Bersih"] > 100000)
]

hasil_filter
```

# Task 5 — Grouping & Agregasi (Pelaporan Kinerja Gerai)

## 1. Laporan Agregasi Per Gerai

Pada bagian ini dibuat laporan ringkasan kinerja setiap gerai. Data terlebih dahulu disaring agar hanya transaksi dengan status `SELESAI` yang digunakan. Selanjutnya, data dikelompokkan berdasarkan `kode_gerai` menggunakan `.groupby()` dan dihitung beberapa metrik menggunakan `.agg()`, yaitu jumlah transaksi, total unit terjual, total omzet bersih, dan rata-rata nilai transaksi. Hasilnya kemudian diurutkan berdasarkan `Total_Omzet_Bersih` dari yang tertinggi ke terendah dan dibulatkan menjadi 2 angka desimal.

```python
laporan_gerai = (
    df[df["status"] == "SELESAI"]
    .groupby("kode_gerai")
    .agg(
        Jumlah_Transaksi=("id_transaksi", "count"),
        Total_Unit_Terjual=("jumlah", "sum"),
        Total_Omzet_Bersih=("Total_Bersih", "sum"),
        Rata_Rata_Transaksi=("Total_Bersih", "mean")
    )
    .sort_values("Total_Omzet_Bersih", ascending=False)
    .round(2)
)

laporan_gerai
```

## 2. Transformasi Kelompok (Window Function)

Pada bagian ini ditambahkan kolom baru bernama `Kontribusi_Gerai_Persen` ke DataFrame utama. Kolom ini menunjukkan persentase kontribusi nilai `Total_Bersih` setiap transaksi terhadap total omzet bersih gerai tempat transaksi tersebut berada. Perhitungannya menggunakan `.groupby().transform()` sehingga total omzet dapat dihitung untuk setiap gerai tanpa mengubah jumlah baris pada DataFrame utama.

```python
total_omzet_gerai = df.groupby("kode_gerai")["Total_Bersih"].transform("sum")

df["Kontribusi_Gerai_Persen"] = (
    df["Total_Bersih"] / total_omzet_gerai * 100
)

df[[
    "id_transaksi",
    "kode_gerai",
    "Total_Bersih",
    "Kontribusi_Gerai_Persen"
]].head(10)
```
